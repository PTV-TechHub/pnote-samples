# Sao chép dữ liệu mẫu vào thư mục của tôi

Thư mục `samples/` chỉ đọc. Notebook này chép các thư mục mẫu sang **`mau-cua-toi/`** trong thư mục gốc của bạn
(`/home/jovyan/work/mau-cua-toi`), nơi bạn sửa và lưu được.

- Chọn **Run → Run All Cells**.
- Chỉ muốn một thư mục? Sửa danh sách `FOLDERS` ở ô dưới (ví dụ `["leanbot-basic"]`).
- Tệp bạn **đã có** trong `mau-cua-toi/` được **giữ nguyên** (không ghi đè). Muốn lấy bản mới, đổi tên hoặc xoá tệp cũ trước.

```python
import shutil
from pathlib import Path

SRC = Path("/home/jovyan/work/samples")
DST = Path("/home/jovyan/work/mau-cua-toi")
FOLDERS = ["leanbot-basic", "digital-twin", "synapse-city"]

copied = kept = 0
for name in FOLDERS:
    for f in sorted((SRC / name).rglob("*")):
        if not f.is_file():
            continue
        target = DST / name / f.relative_to(SRC / name)
        if target.exists():
            kept += 1
            continue
        target.parent.mkdir(parents=True, exist_ok=True)
        shutil.copyfile(f, target)
        copied += 1
print(f"Đã chép {copied} tệp, giữ nguyên {kept} tệp đã có, vào {DST}")
print("Mở thư mục mau-cua-toi/ ở thanh tệp bên trái (có thể cần bấm nút làm mới).")
```
